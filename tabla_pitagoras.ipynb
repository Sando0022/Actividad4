```python
Parte A — Construir la matriz de 10 × 10
tabla = []

for fila in range(1, 11):
    renglon = []
    for columna in range(1, 11):
        renglon.append(fila * columna)
    tabla.append(renglon)

# Mostrar la matriz construida (opcional para comprobar su estructura)
print("Matriz creada: 10 filas y 10 columnas")
print("Cantidad de filas:", len(tabla))
print("Cantidad de columnas:", len(tabla[0]))
```

```python
# Partes B y C — Funciones para imprimir y consultar la tabla

def imprimir_tabla(tabla):
    for fila in tabla:
        for valor in fila:
            print(valor, end="\t")
        print()

def consultar_producto(tabla, renglon, columna):
    return tabla[renglon - 1][columna - 1]

# Imprimir la tabla completa
imprimir_tabla(tabla)
```

```python
# Parte D — Programa principal con captura y validación

renglon = int(input("Ingresa el factor del renglón (1 a 10): "))
while renglon < 1 or renglon > 10:
    print("Aviso: el factor debe estar entre 1 y 10.")
    renglon = int(input("Vuelve a ingresar el factor del renglón (1 a 10): "))

columna = int(input("Ingresa el factor de la columna (1 a 10): "))
while columna < 1 or columna > 10:
    print("Aviso: el factor debe estar entre 1 y 10.")
    columna = int(input("Vuelve a ingresar el factor de la columna (1 a 10): "))

resultado = consultar_producto(tabla, renglon, columna)
print(f"{renglon} x {columna} = {resultado}")
```
